## 2 - Math & Comparisons

```python
a= 10
b=3

print(a + b)      # 13     addition
print(a - b)      # 7      subtraction
print(a * b)      # 30     multiplication
print(a / b)      # 3.333  division — always float
print(a // b)     # 3      floor division — drops decimal
print(a % b)      # 1      remainder
print(a ** 2)     # 100    power
print(a ** 0.5)   # 3.16   square root
```

```python
# Shorthand assignment
n = 0
n += 1    # n = n + 1
n += 1
print(n)  # 2

n *= 3    # n = n * 3
print(n)  # 6
```

```python
# Comparisons — always return True or False
print(5 == 5)    # True
print(5 != 3)    # True
print(5 > 3)     # True
print(5 < 3)     # False
print(5 >= 5)    # True
print(5 <= 4)    # False

# Combine conditions
age = 25
print(age > 18 and age < 60)   # True  — both must be True
print(age < 10 or age > 80)    # False — at least one must be True
print(not (age > 18))          # False — reverses

# Python chain (elegant)
print(18 <= age <= 60)         # True
```

## 7 - Conditions

```python
score = 0.84

# Full if / elif / else
if score >= 0.90:
    print('Excellent')
elif score >= 0.80:
    print('Good')
elif score >= 0.70:
    print('Acceptable')
else:
    print('Needs improvement')

# INDENTATION IS REQUIRED — 4 spaces
```

```python
# One-liner (ternary) — very common in ML!
prob       = 0.73
prediction = 1 if prob >= 0.5 else 0
label      = 'Survived' if prediction == 1 else 'Died'
print(f'P={prob:.2f} -> prediction={prediction} -> {label}')

# Check overfit
train_acc, test_acc = 0.95, 0.72
gap    = train_acc - test_acc
status = 'OVERFITTING!' if gap > 0.1 else 'OK'
print(f'Gap={gap:.2f} -> {status}')
```

```python
# Conditions in functions — feature engineering
def age_group(age):
    if age < 12: 
        return 'Child'
    elif age < 18:
        return 'Teen'
    elif age < 60:
        return 'Adult'
    else:         
        return 'Senior'

def fare_tier(fare):
    if fare < 10: 
        return 'Low'
    elif fare < 50:
        return 'Medium'
    else:        
        return 'High'

test_ages  = [8, 15, 35, 72]
test_fares = [7, 25, 120, 5]

for age, fare in zip(test_ages, test_fares):
    print(f'Age={age:3d} ->> {age_group(age):<7}  |  Fare={fare:4d} -> {fare_tier(fare)}')
```

## 8 - Loops

```python
# for loop over a list
features = ['Age', 'Fare', 'Pclass', 'Sex']

for feat in features:
    print(f'Feature: {feat}')
```

```python
# range() — repeat N times
for i in range(5):
    print(f'Fold {i+1}')
print()

# range(start, stop, step)
for i in range(0, 11, 2):
    print(i, end=' ')
print()
```

```python
# enumerate() — index + value
scores = [0.82, 0.79, 0.84, 0.81, 0.83]

for i, score in enumerate(scores):
    print(f'Fold {i}:: {score:.2f}')
```

```python
# zip() — loop two lists together
features = ['Age', 'Fare', 'Pclass', 'Sex']
coefs    = [0.45, 0.31, -0.62, 0.89]

for feat, coef in zip(features, coefs):
    direction = 'positive' if coef > 0 else 'negative'
    print(f'{feat:8}: {coef:+.2f}  ({direction})')
```

```python
# Collect results in a loop
import random
random.seed(42)

fold_scores = []
for fold in range(5):
    score = random.uniform(0.78, 0.87)   # simulate a score
    fold_scores.append(score)
    print(f'Fold {fold+1}: {score:.4f}')

mean = sum(fold_scores) / len(fold_scores)
print(f'Mean: {mean:.4f}')
```

## 9 - List Comprehensions
> Loop في سطر واحد  هتشوفه كتير في كود ML

```python
# Normal loop way
nums   = [1, 4, 9, 16, 25]
result = []
for n in nums:
    result.append(n ** 0.5)
print(result)

# Same — list comprehension
result = [n ** 0.5 for n in nums]
print(result)
```

```python
# With condition (filter)
scores = [82, 79, 84, 81, 83, 68, 91, 75]

high   = [s for s in scores if s >= 82]
print(f'High scores: {high}')

# Transform
scaled = [s / 100 for s in scores]
print(f'Scaled: {[round(s, 2) for s in scaled]}')

# On strings
cols  = ['Age', 'Fare', 'Pclass', 'Sex']
lower = [c.lower() for c in cols]
print(lower)
```

```python
# Real ML patterns

# Predictions from probabilities
probs = [0.9, 0.3, 0.7, 0.45, 0.88, 0.12]
preds = [1 if p >= 0.5 else 0 for p in probs]
print(f'Probs: {probs}')
print(f'Preds: {preds}')

# Clip values
fares   = [5, 50, 512, 3, 200, 7]
clipped = [min(f, 100) for f in fares]   # cap at 100
print(f'Fares clipped: {clipped}')

# Count correct predictions
y_true = [1, 0, 1, 1, 0, 1]
y_pred = [1, 0, 0, 1, 0, 1]
correct = sum(1 for t, p in zip(y_true, y_pred) if t == p)
print(f'Correct: {correct}/{len(y_true)} = {correct/len(y_true):.1%}')
```

## 10 - Functions

```python
# Basic function
def greet(name):
    print(f'Hello, {name}!')

greet('Jack')
greet('Rose')
```

```python
# With return value
def pct(part, total):
    return part / total * 100

print(f'Survival: {pct(342, 891):.1f}%')
print(f'Female survival: {pct(233, 314):.1f}%')
print(f'Male survival: {pct(109, 577):.1f}%')
```

```python
# Default parameter values
def display(label, value, decimals=2):
    print(f'{label}: {value:.{decimals}f}')

display('Accuracy', 0.8234)         # uses default: 2
display('Accuracy', 0.8234, 4)      # override to 4
display('RMSE',     3.56712, 3)
```

```python
# Return multiple values (returns a tuple)
def stats(numbers):
    n    = len(numbers)
    mean = sum(numbers) / n
    return min(numbers), max(numbers), mean

lo, hi, avg = stats([0.82, 0.79, 0.84, 0.81, 0.83])
print(f'Min={lo:.2f}  Max={hi:.2f}  Mean={avg:.4f}')
```

```python
# lambda — one-line anonymous function
square = lambda x: x ** 2
print(square(5))    # 25
print(square(3))    # 9

# Lambda with map()
scores  = [82, 79, 84, 81]
scaled  = list(map(lambda x: x / 100, scores))
print(scaled)

# Lambda to sort by second element
pairs = [('Age', 0.45), ('Fare', 0.31), ('Sex', 0.89)]
pairs_sorted = sorted(pairs, key=lambda x: x[1], reverse=True)
print(pairs_sorted)
```

```python
# enumerate and zip — used all the time
cols   = ['Age', 'Fare', 'Pclass']
coefs  = [0.45, 0.31, -0.62]

print('With enumerate:')
for i, col in enumerate(cols):
    print(f'  [{i}] {col}')

print('\nWith zip:')
for col, coef in zip(cols, coefs):
    print(f'  {col:8}: {coef:+.2f}')
```

```python
# isinstance — check type at runtime
values = [25, 3.14, 'text', True, None, [1,2,3]]

for v in values:
    if isinstance(v, (int, float)) and not isinstance(v, bool):
        print(f'{str(v):<10} -> numeric')
    elif isinstance(v, str):
        print(f'{str(v):<10} -> string')
    elif isinstance(v, bool):
        print(f'{str(v):<10} -> boolean')
    elif v is None:
        print(f'{str(v):<10} -> None')
    else:
        print(f'{str(v):<10} -> {type(v).__name__}')
```

## Final Practice - Put It All Together

```python
# Simulate what you will write in ML notebooks

# Variables
dataset  = 'Titanic'
n_rows   = 891
n_cols   = 12
survived = 342

# f-string
print(f'Dataset : {dataset}')
print(f'Shape   : {n_rows} x {n_cols}')
print(f'Survived: {survived} ({survived/n_rows:.1%})')
print()

# List of features
features     = ['Age', 'Fare', 'Pclass', 'Sex', 'SibSp', 'Parch']
cols_to_drop = ['Name', 'Ticket', 'Cabin', 'PassengerId']
print(f'Features   : {features}')
print(f'Drop       : {cols_to_drop}')
print()

# Dictionary of results
results = {'train': 0.9123, 'test': 0.8234, 'gap': 0.0889}
for k, v in results.items():
    print(f'  {k:6}: {v:.4f}')
print()

# Condition
status = 'Overfitting!' if results['gap'] > 0.1 else 'Good fit'
print(f'Status: {status}')
print()

# Loop + function
def categorize(score):
    if score >= 0.90:   return 'Excellent'
    elif score >= 0.80: return 'Good'
    elif score >= 0.70: return 'Acceptable'
    else:               return 'Poor'

cv_scores = [0.82, 0.79, 0.84, 0.81, 0.83]
print('CV Results:')
for i, score in enumerate(cv_scores):
    print(f'  Fold {i+1}: {score:.4f}  ({categorize(score)})')
print(f'  Mean  : {sum(cv_scores)/len(cv_scores):.4f}')
print()

# Comprehension
high_folds = [i+1 for i, s in enumerate(cv_scores) if s >= 0.82]
print(f'Folds above 0.82: {high_folds}')
```
