## 1️⃣ ما هي Pandas؟

**Pandas** هي مكتبة Python مخصصة لـ **تحليل البيانات ومعالجتها**.

فكّرها كـ **Excel** بس بقوة Python!

### لماذا Pandas؟
- تقدر تقرأ ملفات CSV, Excel, JSON وغيرها
- تنظّف البيانات وتعالج القيم الناقصة
- تحلّل وتستخرج إحصائيات بسهولة
- أساس لكل مشاريع Machine Learning

### العلاقة مع NumPy
```
NumPy  →  أرقام وعمليات رياضية (arrays)
Pandas →  جداول وبيانات حقيقية (DataFrames)
```

````python
# أول خطوة دائماً: import
import pandas as pd
import numpy as np

print('✅ Pandas version:', pd.__version__)
print('✅ NumPy version:', np.__version__)
````

## 2️⃣ Series – العمود الواحد

**Series** = عمود واحد من البيانات، له **index** (رقم أو اسم لكل عنصر)

```
Index  |  Value
-------|-------
  0    |   85
  1    |   90
  2    |   78
```

````python
# إنشاء Series من قائمة
grades = pd.Series([85, 90, 78, 92, 88])
print('📊 درجات الطلاب:')
print(grades)
print()
print('النوع:', type(grades))
````

````python
# Series بـ index مخصص (اسم بدل رقم)
grades = pd.Series(
    [85, 90, 78, 92, 88],
    index=['أحمد', 'سارة', 'محمد', 'ليلى', 'خالد']
)
print('📊 درجات الطلاب مع الأسماء:')
print(grades)
````

````python
# الوصول لعنصر معين
print('درجة سارة:', grades['سارة'])
print('درجة أحمد:', grades['أحمد'])

# إحصاءات سريعة
print()
print('📈 أعلى درجة:', grades.max())
print('📉 أدنى درجة:', grades.min())
print('📊 المتوسط:', grades.mean())
````

````python
# عمليات على Series (مثل NumPy)
print('🎯 من نجح (درجة > 80):')
print(grades[grades > 80])
````

## 3️⃣ DataFrame – الجدول

**DataFrame** = جدول كامل من الأعمدة والصفوف  
كل عمود هو Series، وكلهم يشتركون في نفس الـ index

```
   الاسم   | العمر | المدينة
-----------|-------|--------
   أحمد    |  25   | الرياض
   سارة    |  30   | جدة
   محمد    |  22   | الدمام
```

````python
# إنشاء DataFrame من dictionary
data = {
    'الاسم':    ['أحمد', 'سارة', 'محمد', 'ليلى', 'خالد'],
    'العمر':    [25, 30, 22, 28, 35],
    'المدينة':  ['الرياض', 'جدة', 'الدمام', 'الرياض', 'جدة'],
    'الراتب':   [8000, 12000, 6500, 9500, 15000]
}

df = pd.DataFrame(data)
print(df)
````

````python
# معلومات أساسية عن الـ DataFrame
print('🔢 الشكل (صفوف × أعمدة):', df.shape)
print('📋 أسماء الأعمدة:', df.columns.tolist())
print('🔑 الـ index:', df.index.tolist())
````

## 7️⃣ التعامل مع القيم الناقصة (Missing Values)

في البيانات الحقيقية، دائماً في قيم ناقصة تظهر كـ **NaN** (Not a Number).  
لازم تعرف كيف تتعامل معها!

````python
# بيانات فيها قيم ناقصة
messy_data = {
    'name':       ['أحمد', 'سارة', 'محمد', 'ليلى', 'خالد'],
    'age':        [25, None, 22, 28, None],
    'salary':     [8000, 12000, None, 9500, 15000],
    'department': ['IT', 'HR', 'IT', None, 'Marketing']
}

df_messy = pd.DataFrame(messy_data).style.background_gradient(cmap='Reds')
print('❗ البيانات مع القيم الناقصة:')
df_messy
````

````python
# كشف القيم الناقصة
print('🔍 أين القيم الناقصة؟')
print(df_messy.isnull())
print()
print('📊 عدد القيم الناقصة في كل عمود:')
print(df_messy.isnull().sum())
````

````python
# الحل 1: حذف الصفوف الناقصة
df_dropped = df_messy.dropna()
print('🗑️ بعد حذف الصفوف الناقصة:')
print(df_dropped)
````

````python
# الحل 2: ملء القيم الناقصة
df_filled = df_messy.copy()

# ملء العمر بالمتوسط
df_filled['age'] = df_filled['age'].fillna(df_filled['age'].mean())

# ملء الراتب بالوسيط (median)
df_filled['salary'] = df_filled['salary'].fillna(df_filled['salary'].median())

# ملء القسم بـ 'Unknown'
df_filled['department'] = df_filled['department'].fillna('Unknown')

print('✅ بعد ملء القيم الناقصة:')
df_filled
````

## 9️⃣ التجميع والإحصاء – groupby()

**groupby** = تجميع البيانات حسب فئة وحساب إحصاءات لكل فئة  
مثل: "ما متوسط الراتب في كل قسم؟"


````python
# متوسط الراتب لكل قسم
avg_salary = employees.groupby('department')['salary'].mean()
print('💼 متوسط الراتب لكل قسم:')
print(avg_salary)
````

````python
# عدد الموظفين في كل قسم
dept_count = employees.groupby('department')['name'].count()
print('👥 عدد الموظفين في كل قسم:')
print(dept_count)
````

````python
# إحصاءات متعددة دفعة وحدة
dept_stats = employees.groupby('department')['salary'].agg(['mean', 'min', 'max', 'count'])
dept_stats.columns = ['متوسط', 'أدنى', 'أعلى', 'عدد الموظفين']
print('📊 إحصاءات الرواتب لكل قسم:')
dept_stats
````

````python
# ترتيب البيانات
print('🏆 الموظفون مرتبون من الأعلى راتباً:')
employees.sort_values('salary', ascending=False)[['name', 'department', 'salary']]
````

## 🔟 Merging & Joining – دمج أكثر من DataFrame

في الواقع، البيانات مش دايماً في جدول واحد.  
مثلاً: بيانات الموظفين في جدول، وبيانات أقسامهم في جدول ثاني.

Pandas عندها طريقتين رئيسيتين للدمج:

| الأمر | الاستخدام |
|-------|----------|
| `pd.merge()` | دمج جدولين بناءً على عمود مشترك (مثل SQL JOIN) |
| `pd.concat()` | لصق جداول فوق بعض أو جنب بعض |

### أنواع الـ merge:
```
inner  →  الصفوف الموجودة في الجدولين معاً فقط
left   →  كل الجدول الأيسر + ما يقابله من الأيمن
right  →  كل الجدول الأيمن + ما يقابله من الأيسر
outer  →  كل الصفوف من الجدولين
```

````python
# جدول 1: بيانات الموظفين
df_employees = pd.DataFrame({
    'employee_id': [101, 102, 103, 104, 105],
    'name':        ['Ahmed', 'Sara', 'Mohamed', 'Layla', 'Khaled'],
    'dept_id':     [1, 2, 1, 3, 1]
})

# جدول 2: بيانات الأقسام
df_departments = pd.DataFrame({
    'dept_id':   [1, 2, 3, 4],
    'dept_name': ['Engineering', 'Marketing', 'HR', 'Finance'],
    'location':  ['Riyadh', 'Jeddah', 'Riyadh', 'Dammam']
})

print('👤 جدول الموظفين:')
print(df_employees)
print()
print('🏢 جدول الأقسام:')
print(df_departments)
````

````python
# INNER JOIN – الموظفون مع أقسامهم (الأقسام الموجودة فقط)
df_inner = pd.merge(df_employees, df_departments, on='dept_id', how='inner')
print('🔗 INNER JOIN – الموظفون مع أقسامهم:')
df_inner
````

````python
# LEFT JOIN – كل الموظفين حتى لو قسمهم مش موجود
df_emp_extra = pd.concat([
    df_employees,
    pd.DataFrame([{'employee_id': 106, 'name': 'Nora', 'dept_id': 99}])
], ignore_index=True)

df_left = pd.merge(df_emp_extra, df_departments, on='dept_id', how='left')
print('⬅️ LEFT JOIN – كل الموظفين (حتى لو قسمهم غير موجود):')
df_left
````

````python
# OUTER JOIN – كل شيء من الجدولين
df_outer = pd.merge(df_emp_extra, df_departments, on='dept_id', how='outer')
print('🔄 OUTER JOIN – كل الموظفين وكل الأقسام:')
df_outer
````

````python
# دمج بأعمدة مختلفة الأسماء
df_salaries = pd.DataFrame({
    'emp_id':  [101, 102, 103, 104, 105],
    'salary':  [12000, 9500, 8500, 7500, 15000]
})

df_merged = pd.merge(
    df_employees,
    df_salaries,
    left_on='employee_id',
    right_on='emp_id',
    how='inner'
)
print('💰 الموظفون مع رواتبهم (أعمدة مختلفة الأسماء):')
df_merged[['employee_id', 'name', 'dept_id', 'salary']]
````

````python
# pd.concat() – لصق جداول فوق بعض
new_employees = pd.DataFrame({
    'employee_id': [111, 112],
    'name':        ['Tariq', 'Hana'],
    'dept_id':     [2, 3]
})

all_employees = pd.concat([df_employees, new_employees], ignore_index=True)
print(f'👥 كل الموظفين بعد الدمج ({len(all_employees)} موظف):')
all_employees
````

````python
# concat جنب بعض (axis=1) – إضافة أعمدة
df_info = pd.DataFrame({'phone': ['055-1111', '055-2222', '055-3333', '055-4444', '055-5555']})
df_full = pd.concat([df_employees, df_info], axis=1)
print('📞 الموظفون مع أرقام هواتفهم:')
df_full
````

### 🧠 متى تستخدم أي أمر؟

```
pd.merge()  ←  عندك عمود مشترك بين جدولين (مثل ID)
pd.concat() ←  تريد تجميع صفوف أو أعمدة من جداول متشابهة
```

## 🎯 Capstone – تحليل بيانات متكامل

الآن نطبّق كل اللي تعلمناه على تحليل حقيقي!

**السيناريو:** عندنا بيانات مبيعات محل، ونريد:
1. تنظيف البيانات
2. إضافة أعمدة محسوبة
3. استخراج تقرير بالمبيعات لكل فئة

````python
# بيانات المبيعات
sales_csv = """order_id,product,category,quantity,unit_price,discount
1,Laptop,Electronics,2,3500,0.10
2,T-Shirt,Clothing,5,120,0
3,Phone,Electronics,1,2200,0.05
4,Jeans,Clothing,3,250,0.15
5,Headphones,Electronics,4,350,0
6,Jacket,Clothing,2,450,0.10
7,Tablet,Electronics,1,1800,0
8,Shoes,Clothing,6,300,0.20
9,Smartwatch,Electronics,3,800,0.05
10,Dress,Clothing,4,200,0
"""

sales = pd.read_csv(io.StringIO(sales_csv))
print('🛒 بيانات المبيعات:')
sales
````

````python
# الخطوة 1: استكشاف البيانات
print('📐 الشكل:', sales.shape)
print('❓ قيم ناقصة:\n', sales.isnull().sum())
````

````python
# الخطوة 2: إضافة أعمدة محسوبة
sales['revenue'] = sales['quantity'] * sales['unit_price']
sales['discount_amount'] = sales['revenue'] * sales['discount']
sales['net_revenue'] = sales['revenue'] - sales['discount_amount']

print('💰 بعد حساب الإيرادات:')
sales[['product', 'quantity', 'unit_price', 'revenue', 'discount_amount', 'net_revenue']]
````

````python
# الخطوة 3: تقرير حسب الفئة
category_report = sales.groupby('category').agg(
    عدد_الطلبات=('order_id', 'count'),
    إجمالي_الكمية=('quantity', 'sum'),
    إجمالي_الإيراد=('net_revenue', 'sum'),
    متوسط_الإيراد=('net_revenue', 'mean')
).round(2)

print('📊 تقرير المبيعات حسب الفئة:')
category_report
````

````python
# الخطوة 4: أفضل 3 منتجات ربحاً
print('🏆 أفضل 3 منتجات من حيث الإيراد:')
sales.sort_values('net_revenue', ascending=False)[['product', 'category', 'net_revenue']].head(3)
````

````python
# الخطوة 5: ملخص نهائي
total_revenue = sales['net_revenue'].sum()
total_orders = len(sales)
avg_order = sales['net_revenue'].mean()

print('=' * 40)
print('         📋 ملخص الأداء التجاري')
print('=' * 40)
print(f'  إجمالي الطلبات:   {total_orders}')
print(f'  إجمالي الإيراد:   {total_revenue:,.0f} ريال')
print(f'  متوسط قيمة الطلب: {avg_order:,.0f} ريال')
print('=' * 40)
````
