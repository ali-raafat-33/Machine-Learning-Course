```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
from google.colab import files

#uploaded = files.upload()          # a dialog will appear — pick result.xlsx

data = pd.read_excel('result.xlsx')
print(data)
```

```python
x = data.drop('result', axis=1)   # all columns except result
y = data['result']               # only the result column

print("X (features):")
print(x.head())
print("\ny (labels):")
print(y.head())
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=0
)

print("Training rows :", len(x_train))
print("Testing rows  :", len(x_test))
```

```python
model_one   = LogisticRegression()
train_model = model_one.fit(x_train, y_train)

print("Model trained successfully!")
```

```python
predict = train_model.predict(x_test)

print("Predicted :", predict)
print("Actual    :", y_test.values)
```

```python
accuracy = accuracy_score(y_test, predict)
print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
cm = confusion_matrix(y_test, predict)
print("Confusion Matrix:")
print(cm)
print("\nTP =", cm[1][1], " TN =", cm[0][0])
print("FP =", cm[0][1], " FN =", cm[1][0])
```
