````python
# Series بـ index مخصص (اسم بدل رقم)
grades = pd.Series(
    [85, 90, 78, 92, 88]
)
print('grades with names ')
print(grades)
````

````python
grades = pd.Series(
    [85, 90, 78, 92, 88],
    index=['Ali', 'Mohamed', 'Sara', 'Ahmed','Shahd']
)
print('grades with names ')
print(grades)
````

````python
print("Ali Grades is : ",grades['Ali'])
print("Ahmed Grades is : ",grades['Ahmed'])

print("Hight Score is :",grades.max())
print("Low Score is :",grades.min())
print("Average Score is :",grades.mean())
````

````python
messy_data = {
    'name':       ['Ahmed', 'Sara', 'Mohamed', 'Lila', 'Khaled'],
    'age':        [25, None, 22, 28, None],
    'salary':     [8000, 12000, None, 9500, 15000],
    'department': ['IT', 'HR', 'IT', None, 'Marketing']
}
df=pd.DataFrame(messy_data)
df
````

````python
print ("where is null value")
print(df.isnull())
````

````python
print(df.isnull().sum())
````

````python
df_dropped = df.dropna()
df_dropped
````

````python
#sol2 fill null value by use mean median and unknown
df_filled = df.copy()

#fill null by "Mean"
df_filled['age'] = df_filled['age'].fillna(df_filled['age'].mean())

# fill null by (median)
df_filled['salary'] = df_filled['salary'].fillna(df_filled['salary'].median())

# fill by 'Unknown'
df_filled['department'] = df_filled['department'].fillna('Unknown')


df_filled
````
